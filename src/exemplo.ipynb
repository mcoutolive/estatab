```python
# Calculo de d de Cohen
import numpy as np

mde_percentual = 0.05 # mínimo efeito detectável
media_controle = 100
desvio_padrao = 20
objetivo_teste = "detectar_aumento"

if objetivo_teste == "detectar_aumento" or objetivo_teste == "detectar_diferenca":
    media_tratamento = (1 + mde_percentual)*media_controle
else:
    media_tratamento = (1 - mde_percentual)*media_controle

cohen_d = (media_tratamento - media_controle) / desvio_padrao
print(cohen_d)
```

```python
# Cálculo de h de Cohen

mde_percentual = 0.10
proporcao_controle = 0.50
proporcao_teste = proporcao_controle*(1 + mde_percentual)

cohen_h = 2 * (np.arcsin(np.sqrt(proporcao_teste)) - np.arcsin(np.sqrt(proporcao_controle)))
print(cohen_h)
```

```python
from statsmodels.stats.power import TTestIndPower

analysis = TTestIndPower()

# parâmetros do experimento
cohen_d = 0.3     # Cohen's d
alpha = 0.05
power = 0.8
ratio = 1.0           # tamanhos iguais entre grupos

# tamanho da amostra por grupo
n = analysis.solve_power(
    effect_size=cohen_d,
    alpha=alpha,
    power=power,
    ratio=ratio
)

print(f"Tamanho da amostra por grupo: {n:.0f}")
```
